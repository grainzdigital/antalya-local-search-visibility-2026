# Antalya Local Search Visibility Benchmark 2026

Reproducible descriptive analysis for the Grainz Digital Open Research v1 dataset.

**Important:** the dataset is a purposeful local-search visibility sample, not a random census or causal ranking-factor study.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

DATA = Path("../data/antalya-local-search-visibility-2026.csv")
df = pd.read_csv(DATA)
df.head()
```

```python
summary = {
    "observations": len(df),
    "sectors": df["sector"].nunique(),
    "districts": df["district"].nunique(),
    "median_rating": df["rating"].median(),
    "median_reviews": df["review_count"].median(),
    "rating_4_5_plus_pct": (df["rating"].dropna() >= 4.5).mean() * 100,
    "rating_4_8_plus_pct": (df["rating"].dropna() >= 4.8).mean() * 100,
    "reviews_100_plus_pct": (df["review_count"].dropna() >= 100).mean() * 100,
}
pd.Series(summary)
```

```python
sector = (df.groupby("sector")
    .agg(n=("business_id","count"),
         avg_rating=("rating","mean"),
         median_reviews=("review_count","median"))
    .sort_values("median_reviews", ascending=False))
sector
```

```python
plot_data = sector.sort_values("median_reviews")
plt.figure(figsize=(9,6))
plt.barh(plot_data.index, plot_data["median_reviews"])
plt.xlabel("Median review count")
plt.ylabel("Sector")
plt.title("Median review count by sector")
plt.tight_layout()
plt.show()
```

```python
rating = df["rating"].dropna()
labels = ["<4.5", "4.5–4.7", "4.8+"]
values = [
    (rating < 4.5).sum(),
    ((rating >= 4.5) & (rating < 4.8)).sum(),
    (rating >= 4.8).sum(),
]
plt.figure(figsize=(7,4.5))
plt.bar(labels, values)
plt.ylabel("Number of listings")
plt.title("Displayed rating distribution")
plt.tight_layout()
plt.show()
```

## Interpretation

The sample is intentionally visibility-biased. High ratings and listing completeness should be read as characteristics observed among surfaced businesses, not as proven causes of ranking.
